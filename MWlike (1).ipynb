### COSMOS

```python
print("ucitavanje cosmos podataka")

cos = Table.read(
    COSMOS_FILE,
    format="fits"
).to_pandas()


#cosmos filteri

cos = cos[
    (cos["zphot"] > 0) &
    (cos["zphot"] < 4) &
    (cos["MassMed"] > 0) &
    (cos["SFR"] > -5)
].copy()


#cosmos velicine

cos["logM"] = cos["MassMed"]
cos["logSFR"] = cos["SFR"]


#sfms filter

cos = cos[
    cos["logSFR"] > -3.5
].copy()


print(
    f"Broj COSMOS objekata nakon filtriranja: {len(cos)}"
)
```

### COSMOS redshift binovi

```python
def get_cosmos_bin(z):

    if z == 0:

        return cos[
            (cos["zphot"] >= 0) &
            (cos["zphot"] < 0.5)
        ].copy()

    elif z == 1:

        return cos[
            (cos["zphot"] >= 0.5) &
            (cos["zphot"] < 1)
        ].copy()

    elif z == 2:

        return cos[
            (cos["zphot"] >= 1) &
            (cos["zphot"] < 2)
        ].copy()

    elif z == 3:

        return cos[
            (cos["zphot"] >= 2) &
            (cos["zphot"] < 4)
        ].copy()
```

### COSMOS SFMS

```python
cosmos_fits = {}


for z in range(4):

    cos_sub = get_cosmos_bin(z)

    if len(cos_sub) < 2:

        print(
            f"Premalo COSMOS objekata za z={z}"
        )

        continue


    X = (
        cos_sub["logM"]
        .values
        .reshape(-1, 1)
    )

    y = cos_sub["logSFR"].values


    huber = HuberRegressor()

    huber.fit(
        X,
        y
    )


    cosmos_fits[z] = huber


    print(
        f"COSMOS SFMS fit z={z}: "
        f"slope = {huber.coef_[0]:.4f}, "
        f"intercept = {huber.intercept_:.4f}"
    )
```

### Ucitavanje, TNG

```python
def load_snapshot(
        snap,
        z,
        max_pages=2000):

    url = (
        f"https://www.tng-project.org/api/TNG100-1/"
        f"snapshots/{snap}/subhalos/"
    )

    params = {
        "fields": "sfr,mass_log_msun"
    }

    all_chunks = []
    page = 0
    first = True


    while url is not None and page < max_pages:

        retries = 0


        #pokusaj ucitavanja stranice

        while retries < 5:

            try:

                if first:

                    r = requests.get(
                        url,
                        params=params,
                        headers=headers,
                        timeout=60
                    )

                    first = False

                else:

                    r = requests.get(
                        url,
                        headers=headers,
                        timeout=60
                    )


                #proverava i prvi i svaki sledeći request

                r.raise_for_status()

                break


            except Exception as e:

                retries += 1

                print(
                    f"\nSnapshot {snap}, "
                    f"z={z}, "
                    f"stranica {page + 1}: "
                    f"pokušaj {retries}/5 nije uspeo:"
                )

                print(e)


                if retries < 5:

                    time.sleep(
                        5 * retries
                    )


        #ako ne uspe ni pet pokusaja

        if retries == 5:

            print(
                f"\nSnapshot {snap}, z={z}: "
                f"5 neuspešnih pokušaja, odustaje i nastavlja dalje"
            )

            break


        #json

        data = r.json()


        #provera rezultata

        if (
            "results" not in data
            or len(data["results"]) == 0
        ):

            print(
                f"\nSnapshot {snap}, z={z}: "
                f"nema više rezultata."
            )

            break


        chunk = pd.DataFrame(
            data["results"]
        )


        #logM

        chunk["logM"] = (
            chunk["mass_log_msun"]
        )


        #logSFR

        chunk["logSFR"] = np.log10(
            np.clip(
                chunk["sfr"],
                1e-5,
                None
            )
        )


        #cuvanje stranice

        all_chunks.append(
            chunk
        )


        page += 1


        print(
            f"\rSnapshot {snap}, z={z}: "
            f"stranica {page}, "
            f"objekata = {len(chunk)}, "
            f"ukupno = "
            f"{sum(len(x) for x in all_chunks)}",
            end=""
        )


        #sledeca stranica

        url = data.get(
            "next",
            None
        )


        #kratka pauza

        time.sleep(0.5)


    print()


    #ako nema podataka

    if len(all_chunks) == 0:

        print(
            f"Nema podataka za snapshot {snap}"
        )

        return pd.DataFrame()


    #spajanje stranica

    df = pd.concat(
        all_chunks,
        ignore_index=True
    )


    #redshift

    df["z"] = z


    #random sample

    df = df.sample(
        frac=SAMPLE_FRAC,
        random_state=42
    )


    print(
        f"\nSnapshot {snap}, z={z}: "
        f"ukupno učitano = {page} stranica"
    )

    print(
        f"Nakon SAMPLE_FRAC={SAMPLE_FRAC}: "
        f"{len(df)} objekata"
    )


    return df
```

### SFR-weighted gas metallicity

```python
def get_gasmetallicitysfrweighted(
        snap,
        subhalo_id,
        max_retries=5):

    url = (
        f"https://www.tng-project.org/api/TNG100-1/"
        f"snapshots/{snap}/subhalos/"
        f"{int(subhalo_id)}/"
    )

    for attempt in range(max_retries):

        try:

            r = requests.get(
                url,
                headers=headers,
                timeout=30
            )

            r.raise_for_status()

            data = r.json()

            if "gasmetallicitysfrweighted" in data:

                return data[
                    "gasmetallicitysfrweighted"
                ]

            else:

                print(
                    f"Nema gasmetallicitysfrweighted: "
                    f"snapshot={snap}, "
                    f"subhalo={subhalo_id}"
                )

                return np.nan

        except Exception as e:

            print(
                f"Greška za subhalo "
                f"{subhalo_id}, "
                f"pokušaj {attempt + 1}: "
                f"{e}"
            )

            time.sleep(
                2 * (attempt + 1)
            )

    return np.nan
```

```python
for snap, z in snapshots.items():

    print()
    print()
    print("=" * 70)
    print(
        f"SNAPSHOT {snap}  |  z ≈ {z}"
    )
    print("=" * 70)


    #ucitavanje snapshotova

    df = load_snapshot(
        snap,
        z
    )


    if len(df) == 0:

        print(
            "snapshot nije ucitan."
        )

        continue


    #tng sfms populacija
    #izbacuje logSFR<-4

    df_sfms = df[
        df["logSFR"] >= -4
    ].copy()


    print(
        f"TNG SFMS objekata: "
        f"{len(df_sfms)}"
    )


    #tng sfms hubber fit

    X = (
        df_sfms["logM"]
        .values
        .reshape(-1, 1)
    )

    y = df_sfms["logSFR"].values


    tng_huber = HuberRegressor()

    tng_huber.fit(
        X,
        y
    )


    print(
        f"TNG SFMS fit z={z}: "
        f"slope = {tng_huber.coef_[0]:.4f}, "
        f"intercept = {tng_huber.intercept_:.4f}"
    )


    #sfms predikcija

    df_sfms["SFMS_fit"] = (
        tng_huber.predict(X)
    )


    #odstupanje od tng sfms

    df_sfms["delta_logSFR"] = (
        df_sfms["logSFR"]
        - df_sfms["SFMS_fit"]
    )


    #upper i lower populacija
    # upper: delta_logSFR>-1
    # lower: delta_logSFR<=-1

    upper = df_sfms[
        df_sfms["delta_logSFR"] > -1.0
    ].copy()


    lower = df_sfms[
        df_sfms["delta_logSFR"] <= -1.0
    ].copy()


    print(
        f"TNG upper: {len(upper)}"
    )

    print(
        f"TNG lower: {len(lower)}"
    )


    #mw maseni opseg
    #|logM-logM_MW|<=0.2

    mw_mass_like = upper[
        np.abs(
            upper["logM"]
            - protoMWmass[z]
        ) <= 0.2
    ].copy()


    #mw populacija
    #upper i MW mass range

    N = len(
        mw_mass_like
    )

    print("+++++++++++++++++++++")
    print("Broj objekata u masenom prozoru 0.2 dex",N)
    print("+++++++++++++++++++++")

    #statistika

    if N > 0:

        mw_logSFR = (
            mw_mass_like["logSFR"]
            .values
        )


        median_logSFR = np.median(
            mw_logSFR
        )


        p16 = np.percentile(
            mw_logSFR,
            16
        )


        p84 = np.percentile(
            mw_logSFR,
            84
        )


    else:

        median_logSFR = np.nan

        p16 = np.nan

        p84 = np.nan


    MW_statistics[z] = {

        "N": N,

        "median_logSFR":
            median_logSFR,

        "p16_logSFR":
            p16,

        "p84_logSFR":
            p84
    }


    #ispis statistike za snapshot

    print()
    print(
        "MW POPULACIJA"
    )

    print(
        f"N = {N}"
    )

    print(
        f"median logSFR = "
        f"{median_logSFR:.3f}"
    )

    print(
        f"16–84% interval = "
        f"[{p16:.3f}, {p84:.3f}]"
    )


    #grafik

    plt.figure(
        figsize=(9, 7)
    )


    #cosmos

    cos_sub = get_cosmos_bin(z)


    plt.scatter(
        cos_sub["logM"],
        cos_sub["logSFR"],
        s=10,
        alpha=0.1,
        label=f"COSMOS z={z}",
        color='#577590'
    )


    #cosmos sfms

    if z in cosmos_fits:

        cos_fit = cosmos_fits[z]

        x_cos = np.linspace(
            cos_sub["logM"].min(),
            cos_sub["logM"].max(),
            200
        )


        y_cos = cos_fit.predict(
            x_cos.reshape(-1, 1)
        )


        plt.plot(
            x_cos,
            y_cos,
            linestyle="-",
            linewidth=2,
            label="COSMOS SFMS",
            c='#43aa8b'
        )


    #tng lower

    plt.scatter(
        lower["logM"],
        lower["logSFR"],
        s=8,
        alpha=0.7,
        color="#90be6d",
        label="TNG lower"
    )


    #tng upper

    plt.scatter(
        upper["logM"],
        upper["logSFR"],
        s=8,
        alpha=0.7,
        color="#f9c74f",
        label="TNG upper"
    )


    #tng sfms

    x_tng = np.linspace(
        df_sfms["logM"].min(),
        df_sfms["logM"].max(),
        200
    )


    y_tng = tng_huber.predict(
        x_tng.reshape(-1, 1)
    )


    plt.plot(
        x_tng,
        y_tng,
        color="#f8961e",
        linestyle="-",
        linewidth=2,
        label="TNG SFMS"
    )


    #mw maseni opseg

    plt.axvline(
        protoMWmass[z] - 0.2,
        color="#f3722c",
        linestyle="--",
        linewidth=1.5
    )


    plt.axvline(
        protoMWmass[z] + 0.2,
        color="#f3722c",
        linestyle="--",
        linewidth=1.5,
        label=r"Proto-MW mass $\pm0.2$ dex"
    )


    #proto mw masa

    plt.axvline(
        protoMWmass[z],
        color="#f3722c",
        linestyle=":",
        linewidth=1
    )


    #mw like galaksije

    plt.scatter(
        mw_mass_like["logM"],
        mw_mass_like["logSFR"],
        marker="^",
        s=60,
        color="#f94144",
        linewidth=0.5,
        label="MW-like"
    )


    #ose

    plt.xlabel(
        r"$\log M_\star\ [M_\odot]$"
    )


    plt.ylabel(
        r"$\log \mathrm{SFR}\ "
        r"[M_\odot\,yr^{-1}]$"
    )


    plt.title(
        f"TNG100-1 + COSMOS, z={z}"
    )


    plt.legend(
        fontsize=9
    )


    plt.tight_layout()

    plt.show()


    #uzima metalicnost za mw like objekte

    print()
    print(
        "gasmetallicity "
        "za MW like objekte..."
    )


    for _, galaxy in (
        mw_mass_like.iterrows()
    ):

        subhalo_id = galaxy["id"]


        Z = get_gasmetallicitysfrweighted(
            snap,
            subhalo_id
            )


        '''CSV:
        logM
        logSFR
        Z=gasmetallicity'''

        MW_population[z].append({

            "logM":
                galaxy["logM"],

            "logSFR":
                galaxy["logSFR"],

            "Z":
                Z
        })


        time.sleep(
            0.1
        )
```

# 3D SFR weighted 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline

# Parametar alpha
param_alpha = [1.0, 0.0]

# Za svaki redshift
for z in range(4):

    # Ucitavanje podataka
    df = pd.read_csv(f"MW_population_z{z}weighted.csv")

    # Masa
    logM = df["logM"].values
    M = 10**logM

    # SFR
    logSFR = df["logSFR"].values
    SFR = 10**logSFR

    # Metalicnost
    Z = df["Z"].values

    # Ukupan broj MW-like objekata
    N = len(df)

    # Priprema grafika
    fig = plt.figure(figsize=(12, 9))
    ax = fig.add_subplot(111, projection="3d")

    # Rezultati za obe vrednosti alpha
    for alpha in param_alpha:

        # Racunanje Np
        Np = M**2 * Z**alpha / SFR

        # Maksimalni Np objekat
        max_index = np.argmax(Np)

        max_Np = Np[max_index]
        max_logM = logM[max_index]
        max_logSFR = logSFR[max_index]
        max_Z = Z[max_index]

        # Gornjih 10% objekata po Np
        n_top10 = max(1, int(np.ceil(0.10 * N)))

        top10_indices = np.argsort(Np)[-n_top10:]

        top10_logM = logM[top10_indices]
        top10_logSFR = logSFR[top10_indices]
        top10_Z = Z[top10_indices]
        top10_Np = Np[top10_indices]

        # Statistika
        print()
        print("=" * 65)
        print(f"z = {z} | alpha = {alpha}")
        print("=" * 65)

        print(f"Ukupan broj MW-like objekata = {N}")
        print(f"Broj objekata u gornjih 10% = {n_top10}")

        print("\nMax-Np objekat")
        print(f"logM       = {max_logM:.4f}")
        print(f"logSFR     = {max_logSFR:.4f}")
        print(f"Z          = {max_Z:.6e}")
        print(f"Np         = {max_Np:.6e}")

        print("\nGornjih 10% po Np")
        print(f"min(Np)    = {top10_Np.min():.6e}")
        print(f"max(Np)    = {top10_Np.max():.6e}")
        print(f"median(Np) = {np.median(top10_Np):.6e}")

        print("\nStatistika svih objekata")
        print(f"min(Np)    = {Np.min():.3e}")
        print(f"max(Np)    = {Np.max():.3e}")
        print(f"median(Np) = {np.median(Np):.3e}")

        # Graficki prikaz
        if alpha == 1.0:

            # Svi objekti - alpha = 1
            scatter = ax.scatter(
                logM,
                logSFR,
                Z,
                c=Np,
                cmap="viridis",
                s=15,
                alpha=0.65,
                label=r"All objects ($\alpha=1$)"
            )

            # Gornjih 10%
            ax.scatter(
                top10_logM,
                top10_logSFR,
                top10_Z,
                s=65,
                facecolors="none",
                edgecolors="black",
                linewidths=1.2,
                label=r"Upper 10% ($\alpha=1$)"
            )

            # Maksimum
            ax.scatter(
                [max_logM],
                [max_logSFR],
                [max_Z],
                s=150,
                marker="*",
                c="red",
                edgecolors="black",
                linewidths=1.0,
                label=r"Maximum ($\alpha=1$)"
            )

        else:

            # Svi objekti - alpha = 0
            scatter_alpha0 = ax.scatter(
                logM,
                logSFR,
                Z,
                c=Np,
                cmap="plasma",
                s=15,
                alpha=0.65,
                marker="^",
                label=r"All objects ($\alpha=0$)"
            )

            # Gornjih 10%
            ax.scatter(
                top10_logM,
                top10_logSFR,
                top10_Z,
                s=90,
                facecolors="none",
                edgecolors="blue",
                linewidths=1.5,
                marker="^",
                label=r"Upper 10% ($\alpha=0$)"
            )

            # Maksimum
            ax.scatter(
                [max_logM],
                [max_logSFR],
                [max_Z],
                s=180,
                marker="*",
                c="cyan",
                edgecolors="black",
                linewidths=1.0,
                label=r"Maximum ($\alpha=0$)"
            )

    # Naslov i ose
    ax.set_title(
        f"MW-like population, z = {z}",
        pad=20
    )

    ax.set_xlabel(r"$\log M_\ast\ [M_\odot]$", labelpad=10)
    ax.set_ylabel(r"$\log SFR\ [M_\odot\,yr^{-1}]$", labelpad=10)
    ax.set_zlabel("Z", labelpad=10)

    # Legenda
    ax.legend(loc="upper left", fontsize=8)

    plt.tight_layout()
    plt.show()
```

# 3D SFR FMR 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline

# Parametar alpha
param_alpha = [1.0, 0.0]

# FMR relacija
def FMR(m, s):
    return (
        8.9
        + 0.37*m
        - 0.14*s
        - 0.19*m**2
        + 0.12*m*s
        - 0.054*s**2
    )


# Za svaki redshift
for z in range(4):

    # Ucitavanje podataka
    df = pd.read_csv(
        f"MW_population_z{z}weighted.csv"
    )

    # Masa
    logM = df["logM"].values
    M = 10**logM

    # SFR
    logSFR = df["logSFR"].values
    SFR = 10**logSFR

    # FMR metalicnost
    m = logM - 10
    s = logSFR

    Z = FMR(m, s)

    # Ukupan broj objekata
    N = len(df)

    # Priprema 3D grafika
    fig = plt.figure(figsize=(12, 9))

    ax = fig.add_subplot(
        111,
        projection="3d"
    )

    # Prolazak kroz obe vrednosti alpha
    for alpha in param_alpha:

        # Racunanje Np
        Np = (
            M**2
            * Z**alpha
            / SFR
        )

        df["Np"] = Np

        # Maksimalni Np objekat
        max_index = np.argmax(Np)

        max_Np = Np[max_index]

        max_logM = logM[max_index]
        max_logSFR = logSFR[max_index]
        max_Z = Z[max_index]

        # Gornjih 10% objekata po Np
        n_top10 = max(
            1,
            int(np.ceil(0.10 * N))
        )

        top10_indices = np.argsort(Np)[-n_top10:]

        top10_logM = logM[top10_indices]
        top10_logSFR = logSFR[top10_indices]
        top10_Z = Z[top10_indices]
        top10_Np = Np[top10_indices]

        # Statistika
        print()
        print("=" * 65)
        print(f"z = {z} | alpha = {alpha}")
        print("=" * 65)

        print(f"Ukupan broj MW-like objekata = {N}")
        print(f"Broj objekata u gornjih 10% = {n_top10}")

        print()
        print("MAX-Np OBJEKAT")

        print(f"logM       = {max_logM:.4f}")
        print(f"logSFR     = {max_logSFR:.4f}")
        print(f"Z (FMR)    = {max_Z:.6e}")
        print(f"Np         = {max_Np:.6e}")

        print()
        print("GORNJIH 10% PO Np")

        print(f"min(Np)    = {top10_Np.min():.6e}")
        print(f"max(Np)    = {top10_Np.max():.6e}")
        print(f"median(Np) = {np.median(top10_Np):.6e}")

        print()
        print("STATISTIKA SVIH OBJEKATA")

        print(f"min(Np)    = {Np.min():.3e}")
        print(f"max(Np)    = {Np.max():.3e}")
        print(f"median(Np) = {np.median(Np):.3e}")

        # Graficki prikaz za alpha = 1
        if alpha == 1.0:

            scatter = ax.scatter(
                logM,
                logSFR,
                Z,
                c=Np,
                cmap="viridis",
                s=15,
                alpha=0.65,
                label=r"All objects ($\alpha=1$)"
            )

            # Gornjih 10%
            ax.scatter(
                top10_logM,
                top10_logSFR,
                top10_Z,
                s=65,
                facecolors="none",
                edgecolors="black",
                linewidths=1.2,
                label=r"Upper 10% ($\alpha=1$)"
            )

            # Maksimum
            ax.scatter(
                [max_logM],
                [max_logSFR],
                [max_Z],
                s=150,
                marker="*",
                c="red",
                edgecolors="black",
                linewidths=1.0,
                label=r"Maximum ($\alpha=1$)"
            )

        # Graficki prikaz za alpha = 0
        else:

            scatter_alpha0 = ax.scatter(
                logM,
                logSFR,
                Z,
                c=Np,
                cmap="plasma",
                s=15,
                alpha=0.65,
                marker="^",
                label=r"All objects ($\alpha=0$)"
            )

            # Gornjih 10%
            ax.scatter(
                top10_logM,
                top10_logSFR,
                top10_Z,
                s=90,
                facecolors="none",
                edgecolors="blue",
                linewidths=1.5,
                marker="^",
                label=r"Upper 10% ($\alpha=0$)"
            )

            # Maksimum
            ax.scatter(
                [max_logM],
                [max_logSFR],
                [max_Z],
                s=180,
                marker="*",
                c="cyan",
                edgecolors="black",
                linewidths=1.0,
                label=r"Maximum ($\alpha=0$)"
            )

    # Naslov i ose
    ax.set_title(
        f"MW-like population, z = {z}",
        pad=20
    )

    ax.set_xlabel(
        r"$\log M_\ast\ [M_\odot]$",
        labelpad=10
    )

    ax.set_ylabel(
        r"$\log SFR\ [M_\odot\,yr^{-1}]$",
        labelpad=10
    )

    ax.set_zlabel(
        "Z (FMR)",
        labelpad=10
    )

    # Dve kolor-skale
    cbar1 = fig.colorbar(
        scatter,
        ax=ax,
        shrink=0.65,
        pad=0.08
    )

    cbar1.set_label(r"$N_p$ ($\alpha=1$)")

    cbar2 = fig.colorbar(
        scatter_alpha0,
        ax=ax,
        shrink=0.65,
        pad=0.16
    )

    cbar2.set_label(r"$N_p$ ($\alpha=0$)")

    # Legenda
    ax.legend(
        loc="upper left",
        fontsize=8
    )

    plt.tight_layout()
    plt.show()
```
