```python
import numpy as np
import os

snapshots = [99, 50, 33, 25]

fields = [
    "SubhaloMassType",
    "SubhaloSFR",
    "SubhaloGasMetallicitySfrWeighted"
]

for snap in snapshots:

    filename = f"TNG_groupcat_data/snapshot_{snap}.npz"

    print("\n" + "=" * 60)
    print(f"SNAPSHOT {snap}")
    print("=" * 60)

    if not os.path.exists(filename):
        print("Fajl ne postoji!")
        continue

    data = np.load(filename)

    # Broj ID-jeva
    ids = data["SubhaloID"]

    print(f"\nBroj subhaloa (ID): {len(ids)}")

    # Provera svakog polja
    lengths = []

    for field in fields:

        array = data[field]

        lengths.append(len(array))

        print(f"{field}:")
        print(f"    Oblik: {array.shape}")
        print(f"    Broj redova: {len(array)}")

    # Provera jednakog broja redova
    if len(set(lengths + [len(ids)])) == 1:

        print("\nSVI NIZOVI IMAJU ISTI BROJ REDOVA.")

    else:

        print("\nGRESKA: BROJ REDOVA SE NE POKLAPA")

    # Provera ID-jeva
    expected_ids = np.arange(len(ids))

    if np.array_equal(ids, expected_ids):

        print("ID-jevi odgovaraju indeksima u katalogu.")

    else:

        print("GRESKA: ID-jevi nisu pravilno poređani")

    # Provera da li postoje duplikati
    if len(np.unique(ids)) == len(ids):

        print("Nema dupliranih ID-jeva.")

    else:

        print("GRESKA: Postoje duplirani ID-jevi.")

    # Provera nedostajućih vrednosti
    for field in fields:

        array = data[field]

        print(
            f"{field} | NaN vrednosti:",
            np.isnan(array).sum()
        )

print("\nKraj")
```

```python
import numpy as np

snapshots = [99, 50, 33, 25]

all_data = {}

for snap in snapshots:

    filename = f"TNG_groupcat_data/snapshot_{snap}.npz"

    loaded = np.load(filename)

    all_data[snap] = {
        key: loaded[key]
        for key in loaded.files
    }

    loaded.close()

    print(f'Snapshot {snap}: učitan')
```

```python
snapshots = [99, 50, 33, 25]

filtered_data = {}

for snap in snapshots:

    print("\n" + "=" * 50)
    print(f"SNAPSHOT {snap}")
    print("=" * 50)

    data = all_data[snap]

    # Ukupan broj objekata pre filtriranja
    N_before = len(data["SubhaloID"])

    # Zvezdana masa (indeks 4)
    stellar_mass = data["SubhaloMassType"][:, 4]

    # Selekcija objekata sa pozitivnom zvezdanom masom
    mask = stellar_mass > 0

    # Broj objekata nakon filtriranja
    N_after = np.sum(mask)

    # Broj izbačenih objekata
    N_removed = N_before - N_after

    # Filtriranje svih polja istom maskom
    filtered_data[snap] = {
        field: data[field][mask]
        for field in [
            "SubhaloID",
            "SubhaloMassType",
            "SubhaloSFR",
            "SubhaloGasMetallicitySfrWeighted"
        ]
    }

    print("Broj objekata pre filtriranja:", N_before)
    print("Broj objekata nakon filtriranja:", N_after)
    print("Broj izbačenih objekata:", N_removed)

    print(
        "Procenat izbačenih objekata:",
        round(100 * N_removed / N_before, 2), "%"
    )

print("\nKraj")
```

```python
import numpy as np

final_data = {}

for snap in snapshots:

    data = filtered_data[snap]

    # Racunanje logSFR
    logSFR = np.full(len(data["SubhaloSFR"]), -np.inf)

    positive_sfr = data["SubhaloSFR"] > 0
    logSFR[positive_sfr] = np.log10(data["SubhaloSFR"][positive_sfr])

    # Uslov: logSFR > -4
    mask = logSFR > -4

    # Primena maske na sva polja
    final_data[snap] = {
        key: data[key][mask]
        for key in data.keys()
    }

    N_before = len(data["SubhaloID"])
    N_after = np.sum(mask)
    N_removed = N_before - N_after

    print("=" * 50)
    print(f"SNAPSHOT {snap}")
    print("=" * 50)
    print(f"Broj objekata pre dodatnog filtriranja: {N_before}")
    print(f"Broj objekata nakon dodatnog filtriranja: {N_after}")
    print(f"Broj izbacenih objekata: {N_removed}")
    print(f"Procenat izbacenih objekata: {100*N_removed/N_before:.2f} %")
```

# SFMS

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import HuberRegressor


protoMWmass = [
    10.68,
    10.48,
    10.28,
    9.98
]


# Veza između snapshot-a i crvenog pomaka
snapshot_redshifts = {
    99: 0,
    50: 1,
    33: 2,
    25: 3
}

# Rezultati
TNG_populations = {}
MW_statistics = {}

for snap, z in snapshot_redshifts.items():

    data = final_data[snap]

    # Izvlacenje zvezdane mase i SFR-a
    stellar_mass = data["SubhaloMassType"][:, 4]
    sfr = data["SubhaloSFR"]

    # Konverzija mase u Msun
    mass_msun = stellar_mass * 1e10 / 0.6774

    # Logaritamske vrednosti
    logM = np.log10(mass_msun)
    logSFR = np.log10(sfr)

    # SFMS fit pomocu Huber regresije
    X = logM.reshape(-1, 1)
    y = logSFR

    huber = HuberRegressor()
    huber.fit(X, y)

    # Odstupanje od SFMS relacije
    logSFR_fit = huber.predict(X)
    delta_logSFR = logSFR - logSFR_fit

    # Podela na upper i lower populaciju
    upper_mask = delta_logSFR > -1
    lower_mask = delta_logSFR <= -1

    upper = pd.DataFrame({
        "id": data["SubhaloID"][upper_mask],
        "logM": logM[upper_mask],
        "logSFR": logSFR[upper_mask],
        "delta_logSFR": delta_logSFR[upper_mask]
    })

    lower = pd.DataFrame({
        "id": data["SubhaloID"][lower_mask],
        "logM": logM[lower_mask],
        "logSFR": logSFR[lower_mask],
        "delta_logSFR": delta_logSFR[lower_mask]
    })

    # Izdvajanje MW-like populacije iz upper populacije
    MW_mass = protoMWmass[z]

    MW_mask = np.abs(upper["logM"] - MW_mass) <= 0.2
    MW_like = upper[MW_mask].copy()

    # Cuvanje rezultata
    TNG_populations[z] = {
        "all": pd.DataFrame({
            "id": data["SubhaloID"],
            "logM": logM,
            "logSFR": logSFR,
            "delta_logSFR": delta_logSFR
        }),
        "upper": upper,
        "lower": lower,
        "mw_mass_like": MW_like,
        "huber": huber
    }

    # Osnovna statistika MW-like populacije
    MW_statistics[z] = {
        "N": len(MW_like),
        "median_logM": MW_like["logM"].median(),
        "median_logSFR": MW_like["logSFR"].median()
    }

    print("=" * 50)
    print(f"SNAPSHOT {snap} | z = {z}")
    print("=" * 50)
    print(f"Ukupan broj galaksija: {len(data['SubhaloID'])}")
    print(f"Upper populacija: {len(upper)}")
    print(f"Lower populacija: {len(lower)}")
    print(f"MW-like populacija: {len(MW_like)}")
    print(f"SFMS nagib: {huber.coef_[0]:.4f}")
    print(f"SFMS odsečak: {huber.intercept_:.4f}")
```

```python
for snap, z in snapshot_redshifts.items():

    data = final_data[snap]

    # Metalicitet iz izvornog TNG kataloga
    Z_weighted = data["SubhaloGasMetallicitySfrWeighted"]

    # Povezivanje metaliciteta sa ID-jevima objekata
    Z_dict = dict(zip(data["SubhaloID"], Z_weighted))

    # Dodavanje metaliciteta u postojece populacije
    for population in ["all", "upper", "lower"]:

        df = TNG_populations[z][population]

        df["Z_weighted"] = df["id"].map(Z_dict)
```

```python
TNG_upper_fit = {}
MWupper = {}

for snap, z in snapshot_redshifts.items():

    upper = TNG_populations[z]["upper"]

    X = upper["logM"].values.reshape(-1, 1)
    y = upper["logSFR"].values

    huber_upper = HuberRegressor()
    huber_upper.fit(X, y)

    logSFR_fit = huber_upper.predict(X)

    delta_logSFR = y - logSFR_fit

    upper_mask_new = delta_logSFR > -1
    lower_mask_new = delta_logSFR <= -1

    upper_new = pd.DataFrame({
        "id": upper["id"].values[upper_mask_new],
        "logM": upper["logM"].values[upper_mask_new],
        "logSFR": upper["logSFR"].values[upper_mask_new],
        "Z_weighted": upper["Z_weighted"].values[upper_mask_new],
        "delta_logSFR": delta_logSFR[upper_mask_new]
    })

    MW_mass = protoMWmass[z]
    MW_mask = np.abs(upper_new["logM"] - MW_mass) <= 0.2
    MW_like = upper_new[MW_mask].copy()

    median_logM = MW_like["logM"].median()
    median_logSFR = MW_like["logSFR"].median()
    median_Z = MW_like["Z_weighted"].median()
    median_M = 10**median_logM
    median_SFR = 10**median_logSFR
    
    lower_new = pd.DataFrame({
        "id": upper["id"].values[lower_mask_new],
        "logM": upper["logM"].values[lower_mask_new],
        "logSFR": upper["logSFR"].values[lower_mask_new],
        "Z_weighted": upper["Z_weighted"].values[lower_mask_new],
        "delta_logSFR": delta_logSFR[lower_mask_new]
    })

    # Cuvanje kataloga MW objekata
    MWupper[z] = MW_like.copy()

    TNG_upper_fit[z] = {
        "all": upper.copy(),
        "upper": upper_new,
        "lower": lower_new,
        "mw_mass_like": MW_like,
        "huber": huber_upper
    }

    print("=" * 50)
    print(f"SNAPSHOT {snap} | z = {z}")
    print("=" * 50)
    print(f"Broj galaksija u novom SFMS fit-u: {len(upper)}")
    print(f"Nova upper populacija: {len(upper_new)}")
    print(f"Nova lower populacija: {len(lower_new)}")
    print(f"Nova MW like populacija: {len(MW_like)}")
    print(f"SFMS nagib: {huber_upper.coef_[0]:.4f}")
    print(f"SFMS odsečak: {huber_upper.intercept_:.4f}")

    print("\nMEDIJANE MW-LIKE POPULACIJE")
    print(f"Median M*: {median_M:.4e} Msun")
    print(f"Median SFR: {median_SFR:.4f} Msun/yr")
    print(f"Median Z_weighted: {median_Z:.6f}")


# Katalozi
MWupper0 = MWupper[0]
MWupper1 = MWupper[1]
MWupper2 = MWupper[2]
MWupper3 = MWupper[3]
```

# Refit i celokupna populacija

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import HuberRegressor

protoMWmass = [10.68, 10.48, 10.28, 9.98]

snapshot_redshifts = {
    99: 0,
    50: 1,
    33: 2,
    25: 3
}

h = 0.6774

# Rezultati
TNG_refit_results = {}
MW_statistics_before = {}
MW_statistics_after = {}


for snap, z in snapshot_redshifts.items():

    data = final_data[snap]

    # POCETNI UZORAK
    stellar_mass = data["SubhaloMassType"][:, 4]
    sfr = data["SubhaloSFR"]
    Z_weighted = data["SubhaloGasMetallicitySfrWeighted"]

    # Konverzija mase u Msun
    mass_msun = stellar_mass * 1e10 / h

    # Validne vrednosti za SFMS
    valid = (
        np.isfinite(mass_msun) &
        (mass_msun > 0) &
        np.isfinite(sfr) &
        (sfr > 0)
    )

    # Logaritamske vrednosti
    logM = np.log10(mass_msun[valid])
    logSFR = np.log10(sfr[valid])

    # Pocetni DataFrame
    df = pd.DataFrame({
        "id": data["SubhaloID"][valid],
        "logM": logM,
        "logSFR": logSFR,
        "Z_weighted": Z_weighted[valid]
    })

  #INICIJALNI SFMS FIT

    X = df["logM"].values.reshape(-1, 1)
    y = df["logSFR"].values

    huber_initial = HuberRegressor()
    huber_initial.fit(X, y)

    df["SFMS_initial"] = huber_initial.predict(X)

    df["delta_initial"] = (
        df["logSFR"] - df["SFMS_initial"]
    )

    # Pocetna klasifikacija
    initial_upper_mask = df["delta_initial"] > -1

    initial_upper = df[initial_upper_mask].copy()
    initial_lower = df[~initial_upper_mask].copy()

    # REFIT NA POCETNOJ UPPER POPULACIJI

    X_upper = initial_upper["logM"].values.reshape(-1, 1)
    y_upper = initial_upper["logSFR"].values

    huber_refit = HuberRegressor()
    huber_refit.fit(X_upper, y_upper)

    # PRIMENA REFITA NA CEO POCETNI UZORAK
    X_all = df["logM"].values.reshape(-1, 1)

    df["SFMS_refit"] = huber_refit.predict(X_all)

    df["delta_refit"] = (
        df["logSFR"] - df["SFMS_refit"]
    )

    # Finalna klasifikacija SVIH galaksija
    final_upper_mask = df["delta_refit"] > -1

    final_upper = df[final_upper_mask].copy()
    final_lower = df[~final_upper_mask].copy()

    #POREĐENJE PROMENE POPULACIJA

    # Koliko galaksija je promenilo populaciju
    initial_to_lower = (
        initial_upper_mask & ~final_upper_mask
    ).sum()

    initial_to_upper = (
        ~initial_upper_mask & final_upper_mask
    ).sum()

    changed = (
        initial_upper_mask != final_upper_mask
    ).sum()

    
    # CUVANJE REZULTATA
    TNG_refit_results[z] = {
        "all": df,
        "initial_upper": initial_upper,
        "initial_lower": initial_lower,
        "final_upper": final_upper,
        "final_lower": final_lower,
        "huber_initial": huber_initial,
        "huber_refit": huber_refit,
        "changed": changed,
        "initial_to_lower": initial_to_lower,
        "initial_to_upper": initial_to_upper
    }


    print("\n" + "=" * 65)
    print(f"SNAPSHOT {snap} | z = {z}")
    print("=" * 65)

    print(f"Ukupan početni uzorak: {len(df)}")

    print("\nINICIJALNA KLASIFIKACIJA")
    print(f"Upper: {len(initial_upper)}")
    print(f"Lower: {len(initial_lower)}")

    print("\nFINALNA KLASIFIKACIJA")
    print(f"Upper: {len(final_upper)}")
    print(f"Lower: {len(final_lower)}")

    print("\nPROMENE PRIPADNOSTI")
    print(f"Upper -> Lower: {initial_to_lower}")
    print(f"Lower -> Upper: {initial_to_upper}")
    print(f"Ukupno promenilo populaciju: {changed}")

    print("\nSFMS PARAMETRI")
    print(
        f"Inicijalni fit: "
        f"a = {huber_initial.coef_[0]:.4f}, "
        f"b = {huber_initial.intercept_:.4f}"
    )
    print(
        f"Refit: "
        f"a = {huber_refit.coef_[0]:.4f}, "
        f"b = {huber_refit.intercept_:.4f}"
    )
```

```python
import matplotlib.pyplot as plt
import numpy as np

for z in sorted(TNG_refit_results.keys()):

    results = TNG_refit_results[z]

    df_plot = results["all"]
    final_upper = results["final_upper"]
    final_lower = results["final_lower"]

    huber_refit = results["huber_refit"]

    x_fit = np.linspace(
        df_plot["logM"].min(),
        df_plot["logM"].max(),
        300
    )

    y_fit = (
        huber_refit.coef_[0] * x_fit
        + huber_refit.intercept_
    )

    plt.figure(figsize=(9, 6))

    plt.scatter(
        final_upper["logM"],
        final_upper["logSFR"],
        s=1,
        alpha=0.35,
        label="Active population",
        color="#0b5351"
    )

    plt.scatter(
        final_lower["logM"],
        final_lower["logSFR"],
        s=1,
        alpha=0.35,
        label="Reduced population",
        color="#00a9a5"
    )

    plt.plot(
        x_fit,
        y_fit,
        color="#092327",
        linewidth=2,
        label="Refitted SFMS"
    )

    plt.xlabel(r"$\log(M_*/M_\odot)$")
    plt.ylabel(r"$\log(\mathrm{SFR}/M_\odot\,\mathrm{yr}^{-1})$")
    plt.title(f"TNG100: SFMS refit, z = {z}")

    plt.legend()
    plt.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()
```

```python
protoMWmass = [10.68, 10.48, 10.28, 9.98]

snapshot_redshifts = {
    99: 0,
    50: 1,
    33: 2,
    25: 3
}

h = 0.6774

mass_window_counts = {}

galaxies_in_mass_window = {}

for snap, z in snapshot_redshifts.items():

    data = final_data[snap]

    stellar_mass = data["SubhaloMassType"][:, 4]

    mass_msun = stellar_mass * 1e10 / h

    logM = np.full(len(mass_msun), np.nan)

    valid_mass = (
        np.isfinite(mass_msun) &
        (mass_msun > 0)
    )

    logM[valid_mass] = np.log10(mass_msun[valid_mass])

    MW_mass = protoMWmass[z]

    lower_limit = MW_mass - 0.2
    upper_limit = MW_mass + 0.2

    mass_mask = (
        valid_mass &
        (logM >= lower_limit) &
        (logM <= upper_limit)
    )

    galaxies_in_mass_window[z] = data["SubhaloID"][mass_mask]
    mass_window_counts[z] = np.sum(mass_mask)

    print(f"SNAPSHOT {snap} | z = {z}")
    print(f"Broj galaksija unutar masenog prozora: {mass_window_counts[z]}")

    plt.figure()

    plt.scatter(
        logM[valid_mass],
        np.log10(data["SubhaloSFR"][valid_mass]),
        s=3,
        alpha=0.3,
        color="#e9c46a",
        label="All galaxies"
    )

    plt.scatter(
        logM[mass_mask],
        np.log10(data["SubhaloSFR"][mass_mask]),
        s=3,
        alpha=0.8,
        color="#2a9d8f",
        label="Galaxies within mass window"
    )

    plt.axvline(
        lower_limit,
        color="#264653",
        linestyle="--",
        linewidth=1.5,
        label=r"$\log M_{\mathrm{protoMW}} \pm 0.2$ dex"
    )

    plt.axvline(
        upper_limit,
        color="#264653",
        linestyle="--",
        linewidth=1.5
    )

    plt.axvline(
        MW_mass,
        color="#264653",
        linewidth=1.5,
        label="Proto-MW mass"
    )

    plt.xlabel(r"$\log(M_*/M_\odot)$")
    plt.ylabel(r"$\log(\mathrm{SFR}/M_\odot\,\mathrm{yr}^{-1})$")
    plt.title(f"TNG100: Stellar mass selection, z = {z}")

    plt.legend()
    plt.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()
```

```python
active_MW_counts = {}

active_MW_galaxies = {}

for snap, z in snapshot_redshifts.items():

    results = TNG_refit_results[z]

    df = results["all"]
    final_upper = results["final_upper"]
    final_lower = results["final_lower"]
    huber_refit = results["huber_refit"]

    MW_mass = protoMWmass[z]
    lower_limit = MW_mass - 0.2
    upper_limit = MW_mass + 0.2

    mass_mask = (
        (df["logM"] >= lower_limit) &
        (df["logM"] <= upper_limit)
    )

    active_mask = df["delta_refit"] > -1

    intersection_mask = mass_mask & active_mask

    intersection = df[intersection_mask].copy()

    active_MW_counts[z] = len(intersection)
    active_MW_galaxies[z] = intersection["id"].values

    N_active = active_mask.sum()
    N_reduced = (~active_mask).sum()
    N_mass = mass_mask.sum()
    N_intersection = intersection_mask.sum()

    print("\n" + "=" * 60)
    print(f"SNAPSHOT {snap} | z = {z}")
    print("=" * 60)

    print(f"Aktivna populacija: {N_active}")
    print(f"Redukovana populacija: {N_reduced}")
    print(f"Sve galaksije u masenom prozoru: {N_mass}")
    print(f"Aktivne galaksije u masenom prozoru: {N_intersection}")


    x_fit = np.linspace(df["logM"].min(), df["logM"].max(), 300)

    y_fit = (
        huber_refit.coef_[0] * x_fit
        + huber_refit.intercept_
    )

    plt.figure(figsize=(10, 7))

    plt.scatter(
        final_lower["logM"],
        final_lower["logSFR"],
        s=3,
        alpha=0.3,
        color="#00a9a5",
        label="Reduced population",
        zorder=1
    )

    plt.scatter(
        final_upper["logM"],
        final_upper["logSFR"],
        s=3,
        alpha=0.35,
        color="#0b5351",
        label="Active population",
        zorder=2
    )

    plt.plot(
        x_fit,
        y_fit,
        color="#092327",
        linewidth=2,
        label="Refitted SFMS",
        zorder=3
    )

    plt.axvspan(
        lower_limit,
        upper_limit,
        color="#e9c46a",
        alpha=0.2,
        label="Proto-MW mass window",
        zorder=0
    )

    plt.axvline(
        MW_mass,
        color="#e76f51",
        linestyle="--",
        linewidth=1.5,
        label="Proto-MW mass",
        zorder=3
    )

    plt.scatter(
        intersection["logM"],
        intersection["logSFR"],
        s=3,
        color="#e76f51",
        linewidths=0.5,
        label="Active galaxies within mass window",
        zorder=5
    )

    plt.xlabel(r"$\log(M_*/M_\odot)$")
    plt.ylabel(r"$\log(\mathrm{SFR}/M_\odot\,\mathrm{yr}^{-1})$")
    plt.title(f"TNG100: SFMS refit and proto-MW mass selection, z = {z}")

    plt.legend()
    plt.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()
```

# TNG I COSMOS GRAFICI

```python
import requests
import pandas as pd
import numpy as np
from sklearn.linear_model import HuberRegressor
import matplotlib.pyplot as plt
from astropy.table import Table
import os
import time

COSMOS_FILE = "asu.fit"

protoMWmass = [
    10.68,
    10.48,
    10.28,
    9.98
]

print("ucitavanje cosmos podataka")

cos = Table.read(
    COSMOS_FILE,
    format="fits"
).to_pandas()


#cosmos filteri

cos = cos[
    (cos["zphot"] > 0) &
    (cos["zphot"] < 4) &
    (cos["MassMed"] > 0) &
    (cos["SFR"] > -5)
].copy()


#cosmos velicine

cos["logM"] = cos["MassMed"]
cos["logSFR"] = cos["SFR"]


#sfms filter

cos = cos[
    #cos["logSFR"] > -3.5
    cos["logSFR"] > -4
].copy()


print(
    f"Broj COSMOS objekata nakon filtriranja: {len(cos)}"
)

def get_cosmos_bin(z):

    if z == 0:

        return cos[
            (cos["zphot"] >= 0) &
            (cos["zphot"] < 0.5)
        ].copy()

    elif z == 1:

        return cos[
            (cos["zphot"] >= 0.5) &
            (cos["zphot"] < 1)
        ].copy()

    elif z == 2:

        return cos[
            (cos["zphot"] >= 1) &
            (cos["zphot"] < 2)
        ].copy()

    elif z == 3:

        return cos[
            (cos["zphot"] >= 2) &
            (cos["zphot"] < 4)
        ].copy()

cosmos_fits = {}


for z in range(4):

    cos_sub = get_cosmos_bin(z)

    if len(cos_sub) < 2:

        print(
            f"Premalo COSMOS objekata za z={z}"
        )

        continue


    X = (
        cos_sub["logM"]
        .values
        .reshape(-1, 1)
    )

    y = cos_sub["logSFR"].values


    huber = HuberRegressor()

    huber.fit(
        X,
        y
    )


    cosmos_fits[z] = huber


    print(
        f"COSMOS SFMS fit z={z}: "
        f"slope = {huber.coef_[0]:.4f}, "
        f"intercept = {huber.intercept_:.4f}"
    )

COSMOS_populations = {}
COSMOS_MW_statistics = {}


for z in cosmos_fits.keys():

    print()
    print("=" * 70)
    print(f"COSMOS | z ≈ {z}")
    print("=" * 70)


    cos_sub = get_cosmos_bin(z).copy()


    X = cos_sub["logM"].values.reshape(-1, 1)
    y = cos_sub["logSFR"].values

    cosmos_huber = HuberRegressor()
    cosmos_huber.fit(X, y)

    cos_sub["SFMS_fit"] = cosmos_huber.predict(X)


    cos_sub["delta_logSFR"] = (
        cos_sub["logSFR"] - cos_sub["SFMS_fit"]
    )


    upper = cos_sub[
        cos_sub["delta_logSFR"] > -1.0
    ].copy()

    lower = cos_sub[
        cos_sub["delta_logSFR"] <= -1.0
    ].copy()


    MW_mass = protoMWmass[z]

    mw_mass_like = upper[
        np.abs(
            upper["logM"] - MW_mass
        ) <= 0.2
    ].copy()


    N = len(mw_mass_like)



    mw_logSFR = mw_mass_like["logSFR"].values

    median_logSFR = np.median(mw_logSFR)

    p16 = np.percentile(mw_logSFR, 16)

    p84 = np.percentile(mw_logSFR, 84)


    

    COSMOS_populations[z] = {

        "all": cos_sub,

        "upper": upper,

        "lower": lower,

        "mw_mass_like": mw_mass_like,

        "huber": cosmos_huber
    }


    COSMOS_MW_statistics[z] = {

        "N": N,

        "median_logSFR": median_logSFR,

        "p16_logSFR": p16,

        "p84_logSFR": p84
    }
    
    
    
    print(f"Ukupan broj COSMOS galaksija: {len(cos_sub)}")

    print(f"Upper populacija: {len(upper)}")

    print(f"Lower populacija: {len(lower)}")

    print("-" * 50)

    print(f"MW-like populacija: N = {N}")

    print(f"Median logSFR = {median_logSFR:.4f}")

    print(f"16–84% interval = [{p16:.4f}, {p84:.4f}]")

    print("-" * 50)

    print(f"SFMS nagib: {cosmos_huber.coef_[0]:.4f}")

    print(f"SFMS odsečak: {cosmos_huber.intercept_:.4f}")


    plt.figure(figsize=(9, 7))


    plt.scatter(
        cos_sub["logM"],
        cos_sub["logSFR"],
        s=8,
        alpha=0.15,
        color="#577590",
        label="COSMOS"
    )


    plt.scatter(
        lower["logM"],
        lower["logSFR"],
        s=8,
        alpha=0.6,
        color="#90be6d",
        label="COSMOS lower"
    )


    plt.scatter(
        upper["logM"],
        upper["logSFR"],
        s=8,
        alpha=0.6,
        color="#f9c74f",
        label="COSMOS upper"
    )


    x_cos = np.linspace(
        cos_sub["logM"].min(),
        cos_sub["logM"].max(),
        200
    )

    y_cos = cosmos_huber.predict(
        x_cos.reshape(-1, 1)
    )

    plt.plot(
        x_cos,
        y_cos,
        color="#43aa8b",
        linewidth=2,
        label="COSMOS SFMS"
    )


    plt.axvline(
        MW_mass - 0.2,
        color="#f3722c",
        linestyle="--",
        linewidth=1.5
    )

    plt.axvline(
        MW_mass + 0.2,
        color="#f3722c",
        linestyle="--",
        linewidth=1.5,
        label=r"Proto-MW mass $\pm 0.2$ dex"
    )


    plt.axvline(
        MW_mass,
        color="#f3722c",
        linestyle=":",
        linewidth=1
    )


    plt.scatter(
        mw_mass_like["logM"],
        mw_mass_like["logSFR"],
        marker="^",
        s=60,
        color="#f94144",
        linewidth=0.5,
        label="MW-like"
    )


    plt.xlabel(r"$\log M_\star\ [M_\odot]$")

    plt.ylabel(r"$\log \mathrm{SFR}\ "r"[M_\odot\,yr^{-1}]$")

    plt.title(f"COSMOS SFMS, z={z}")

    plt.legend(fontsize=9)

    plt.tight_layout()

    plt.show()
```
